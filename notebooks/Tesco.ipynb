```python
# handy function from the article
def haversine_vectorise(longitude_x, latitude_x, longitude_y, latitude_y):
    """Returns distance, in kilometers, between one set of longitude/latitude coordinates and another"""
    longitude_x, latitude_x, longitude_y, latitude_y = map(np.radians, [longitude_x, latitude_x, longitude_y, latitude_y])
 
    newlon = longitude_x - longitude_y
    newlat = latitude_x - latitude_y
 
    haver_formula = np.sin(newlat/2.0)**2 + np.cos(latitude_x) * np.cos(latitude_y) * np.sin(newlon/2.0)**2
 
    dist = 2 * np.arcsin(np.sqrt(haver_formula ))
    km = 3958 * dist #6367 for distance in KM for miles use 3958
    return km
```
